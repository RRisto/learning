##### 🤖Reply🤖<!-- SOLVEIT_SEPARATOR_7f3a9b2c -->

Great question! Let me explain the key differences:

**BiasNorm** (what you have):
- Normalizes to mean=0, std=1 (like LayerNorm)
- Then applies learnable scale and bias
- More complex computation

**BasicNorm** (icefall's choice):
- Computes RMS (root mean square) norm of the input
- Normalizes by dividing by that RMS
- Then applies learnable scale and bias
- Simpler, faster computation

The icefall authors found that BasicNorm works just as well as LayerNorm/BiasNorm but is computationally cheaper. It doesn't force mean=0, just normalizes the magnitude.

Think of it this way: BiasNorm says "make the distribution standard normal, then scale it", while BasicNorm says "just normalize the magnitude, then scale it".

````
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchaudio
from torch.utils.data import Dataset, DataLoader

class BasicNorm(nn.Module):
    def __init__(self, num_channels: int, channel_dim: int = -1, eps: float = 0.25):
        super().__init__()
        self.num_channels = num_channels
        self.channel_dim = channel_dim
        self.eps = eps
        self.scale = nn.Parameter(torch.ones(num_channels))
        self.bias = nn.Parameter(torch.zeros(num_channels))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # Compute RMS norm
        rms = torch.sqrt(torch.mean(x ** 2, dim=self.channel_dim, keepdim=True) + self.eps)
        # Normalize
        x = x / rms
        # Apply scale and bias
        # Need to reshape scale/bias to broadcast correctly
        if self.channel_dim == -1:
            return x * self.scale + self.bias
        else:
            # Add dimensions for broadcasting
            shape = [1] * x.ndim
            shape[self.channel_dim] = self.num_channels
            scale = self.scale.view(shape)
            bias = self.bias.view(shape)
            return x * scale + bias
````

````
#my initial code: import torch
import torch.nn as nn
import torch.nn.functional as F
import torchaudio
from torch.utils.data import Dataset, DataLoader

# ============================================
# BASIC BUILDING BLOCKS
# ============================================

def ScaledLinear(*args, initial_scale: float = 1.0, **kwargs) -> nn.Linear:
    ans = nn.Linear(*args, **kwargs)
    with torch.no_grad():
        ans.weight[:] *= initial_scale
        if ans.bias is not None:
            torch.nn.init.uniform_(ans.bias, -0.1 * initial_scale, 0.1 * initial_scale)
    return ans

def swooshr(x):
    return torch.log(1+torch.exp(x-1))-0.08*x-0.313261687

def swooshl(x):
    return torch.log(1+torch.exp(x-4))-0.08*x-0.035

# ============================================
# BIASNORM
# ============================================

class BiasNormFunction(torch.autograd.Function):
    @staticmethod
    def forward(ctx, x, bias, log_scale):
        x_debiased = x - bias
        var = torch.mean(x_debiased**2, dim=-1, keepdim=True)
        scale = var**(-0.5) * torch.exp(log_scale)
        output = x * scale
        ctx.save_for_backward(output, bias, scale, log_scale)
        return output

    @staticmethod
    def backward(ctx, grad_output):
        output, bias, scale, log_scale = ctx.saved_tensors
        x = output / scale
        x_debiased = x - bias
        var = torch.mean(x_debiased**2, dim=-1, keepdim=True)
        
        grad_log_scale = torch.sum(grad_output * output).view(1)
        N = x_debiased.shape[-1]
        common_term = torch.sum(grad_output * x, dim=-1, keepdim=True)
        grad_bias = torch.sum(common_term * scale * var**(-1) * (1/N) * x_debiased, dim=(0,1))
        grad_x = grad_output * scale - scale * var**(-1) * (1/N) * x_debiased * torch.sum(grad_output * x, dim=-1, keepdim=True)
        
        return grad_x, grad_bias, grad_log_scale

class BiasNorm(nn.Module):
    def __init__(self, num_channels):
        super().__init__()
        self.num_channels = num_channels
        self.bias = nn.Parameter(torch.empty(self.num_channels).uniform_(-0.01, 0.01))
        self.log_scale = nn.Parameter(torch.tensor(0.0))

    def forward(self, x):
        return BiasNormFunction.apply(x, self.bias, self.log_scale)

# ============================================
# POSITIONAL ENCODING
# ============================================

class RelPositionalEncoding(nn.Module):
    def __init__(self, embedding_dim, max_pos):
        super().__init__()
        self.embedding_dim = embedding_dim
        self.max_pos = max_pos
        self.emb = nn.Embedding(2*self.max_pos+1, embedding_dim)
    
    def forward(self, x):
        batch_n, seq_len, channels = x.shape
        t = torch.arange(seq_len)
        dist_mat = (t - t.unsqueeze(1)).clip(-self.max_pos, self.max_pos) + self.max_pos
        return self.emb(dist_mat)

# ============================================
# ATTENTION MODULES
# ============================================

class RelPositionMultiheadAttentionWeights(nn.Module):
    def __init__(self, embed_dim, num_heads, query_head_dim, pos_dim, pos_head_dim, dropout=0.0):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.pos_dim = pos_dim
        self.pos_head_dim = pos_head_dim
        self.query_head_dim = query_head_dim
        self.dropout = nn.Dropout(dropout)
        
        #self.in_proj = ScaledLinear(self.embed_dim, (2*self.num_heads*self.query_head_dim) + (self.num_heads*self.pos_head_dim))
        self.in_proj = ScaledLinear(self.embed_dim, (2*self.num_heads*self.query_head_dim) + (self.num_heads*self.pos_dim))

    def forward(self, x, pos_emb):
        batch_n, seq_len, dim = x.shape
        x = self.in_proj(x)
        q = x[..., :self.num_heads*self.query_head_dim]
        k = x[..., self.num_heads*self.query_head_dim:2*self.num_heads*self.query_head_dim]
        p = x[..., 2*self.num_heads*self.query_head_dim:]
        
        q = q.reshape(batch_n, seq_len, self.num_heads, self.query_head_dim).transpose(1, 2)
        k = k.reshape(batch_n, seq_len, self.num_heads, self.query_head_dim).transpose(1, 2)
        #p = p.reshape(batch_n, seq_len, self.num_heads, self.pos_head_dim).transpose(1, 2)
        p = p.reshape(batch_n, seq_len, self.num_heads, self.pos_dim).transpose(1, 2)
        print(f'p shape {p.shape}')
        print(f"pos_emb shape: {pos_emb.shape}")
        pos_scores = torch.einsum('bhid,ijd->bhij', p, pos_emb)
        attn_scores = q @ k.transpose(-2, -1)
        attn_scores = attn_scores/(self.query_head_dim ** 0.5) + pos_scores
        attn_weights = torch.softmax(attn_scores, dim=-1)
        attn_weights = self.dropout(attn_weights)
        return attn_weights

class SelfAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, value_head_dim):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.value_head_dim = value_head_dim
        self.proj1 = ScaledLinear(self.embed_dim, self.num_heads * self.value_head_dim)
        self.proj2 = ScaledLinear(self.num_heads * self.value_head_dim, self.embed_dim)

    def forward(self, x, attn_weights):
        x = self.proj1(x)
        batch_n, seq_len, _ = x.shape
        x = x.reshape((batch_n, seq_len, self.num_heads, self.value_head_dim)).transpose(1, 2)
        x = attn_weights @ x
        x = x.transpose(1, 2).reshape((batch_n, seq_len, -1))
        x = self.proj2(x)
        return x

class NonLinearAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, head_dim):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = head_dim
        self.proj_value1 = ScaledLinear(self.embed_dim, self.num_heads * self.head_dim)
        self.proj_value2 = ScaledLinear(self.num_heads * self.head_dim, self.embed_dim)
        self.proj_query1 = ScaledLinear(self.embed_dim, self.num_heads * self.head_dim)
        self.activation_fun = swooshr

    def forward(self, x, attn_weights):
        value = self.proj_value1(x)
        query = self.proj_query1(x)
        batch_n, seq_len, _ = value.shape
        value = value.reshape((batch_n, seq_len, self.num_heads, self.head_dim)).transpose(1, 2)
        query = query.reshape((batch_n, seq_len, self.num_heads, self.head_dim)).transpose(1, 2)
        output = self.activation_fun(query) * (attn_weights @ value)
        output = output.transpose(1, 2).reshape((batch_n, seq_len, -1))
        output = self.proj_value2(output)
        return output

# ============================================
# FEEDFORWARD AND CONVNEXT
# ============================================

class FeedforwardModule(nn.Module):
    def __init__(self, embed_dim, feedforward_dim, dropout):
        super().__init__()
        self.embed_dim = embed_dim
        self.feedforward_dim = feedforward_dim
        self.dropout = dropout
        self.linear1 = nn.Linear(self.embed_dim, self.feedforward_dim)
        self.activation_fun = swooshl
        self.drop = nn.Dropout(self.dropout)
        self.linear2 = nn.Linear(self.feedforward_dim, self.embed_dim)

    def forward(self, x):
        x = self.linear1(x)
        x = self.activation_fun(x)
        x = self.drop(x)
        x = self.linear2(x)
        return x

class ConvNeXt(nn.Module):
    def __init__(self, embed_dim, kernel_size, expansion_factor, dropout):
        super().__init__()
        self.embed_dim = embed_dim
        self.kernel_size = kernel_size
        self.expansion_factor = expansion_factor
        self.dropout_n = dropout
        
        self.depth_conv = nn.Conv1d(self.embed_dim, self.embed_dim, self.kernel_size, 
                                     padding=self.kernel_size // 2, groups=self.embed_dim)
        self.point_expansion = nn.Linear(self.embed_dim, self.embed_dim*self.expansion_factor)
        self.activation_fun = swooshl
        self.dropout = nn.Dropout(self.dropout_n)
        self.point_compression = nn.Linear(self.embed_dim*self.expansion_factor, self.embed_dim)

    def forward(self, x):
        x = torch.transpose(x, 2, 1)
        x = self.depth_conv(x)
        x = torch.transpose(x, 2, 1)
        x = self.point_expansion(x)
        x = self.activation_fun(x)
        x = self.dropout(x)
        x = self.point_compression(x)
        return x

# ============================================
# ZIPFORMER BLOCK AND ENCODER
# ============================================

class ZipformerBlock(nn.Module):
    def __init__(self, embed_dim, num_heads, query_head_dim, value_head_dim, 
                 pos_dim, pos_head_dim, feedforward_dim, kernel_size, expansion_factor, dropout=0.1):
        super().__init__()
        self.embed_dim = embed_dim
        
        self.attn_weights = RelPositionMultiheadAttentionWeights(
            embed_dim, num_heads, query_head_dim, pos_dim, pos_head_dim, dropout
        )
        self.sa = SelfAttention(embed_dim, num_heads, value_head_dim)
        self.nla = NonLinearAttention(embed_dim, num_heads, value_head_dim)
        self.ffn = FeedforwardModule(embed_dim, feedforward_dim, dropout)
        self.convnext = ConvNeXt(embed_dim, kernel_size, expansion_factor, dropout)
        
        self.norm1 = BiasNorm(embed_dim)
        self.norm2 = BiasNorm(embed_dim)
        self.norm3 = BiasNorm(embed_dim)
        self.norm4 = BiasNorm(embed_dim)

    def forward(self, x, pos_emb):
        attn_weights = self.attn_weights(x, pos_emb)
        x = x + self.sa(self.norm1(x), attn_weights)
        x = x + self.nla(self.norm2(x), attn_weights)
        x = x + self.ffn(self.norm3(x))
        x = x + self.convnext(self.norm4(x))
        return x

class ZipformerEncoder(nn.Module):
    def __init__(self, num_layers, embed_dim=64, num_heads=4, query_head_dim=16, value_head_dim=16,
                 pos_dim=32, pos_head_dim=8, feedforward_dim=256, kernel_size=7,
                 expansion_factor=2, dropout=0.1, embedding_dim=8, max_pos=32):
        super().__init__()
        self.num_layers = num_layers
        self.pos_encoder = RelPositionalEncoding(embedding_dim=embedding_dim, max_pos=max_pos)
        self.layers = nn.ModuleList([ZipformerBlock(
                            embed_dim=embed_dim, num_heads=num_heads, query_head_dim=query_head_dim, 
                            value_head_dim=value_head_dim, pos_dim=pos_dim, pos_head_dim=pos_head_dim, 
                            feedforward_dim=feedforward_dim, kernel_size=kernel_size,
                            expansion_factor=expansion_factor, dropout=dropout
                            ) for i in range(self.num_layers)])

    def forward(self, x):
        pos_emb = self.pos_encoder(x)
        for layer in self.layers:
            x = layer(x, pos_emb)
        return x

# ============================================
# MULTI-SCALE ARCHITECTURE
# ============================================

class Downsample2(nn.Module):
    def __init__(self):
        super().__init__()
        self.weights = nn.Parameter(torch.zeros(2))

    def forward(self, x):
        batch, seq_len, channels = x.shape
        x = x.reshape(batch, seq_len//2, 2, channels)
        normalized_weights = torch.softmax(self.weights, dim=0)
        out = torch.einsum('bpdc,d->bpc', x, normalized_weights)
        return out

class Upsample2(nn.Module):
    def __init__(self):
        super().__init__()
    
    def forward(self, x):
        x = torch.repeat_interleave(x, repeats=2, dim=1)
        return x

class MultiScaleZipformerEncoder(nn.Module):
    def __init__(self, stack_config):
        super().__init__()
        self.stack_config = stack_config
        self.layers = nn.ModuleList([ZipformerEncoder(l['num_layers'], l["embed_dim"], l["num_heads"], 
                    l["query_head_dim"], l["value_head_dim"], l["pos_dim"], l["pos_head_dim"], 
                    l["feedforward_dim"], l["kernel_size"], l["expansion_factor"], l["dropout"],
                    l["embedding_dim"], l["max_pos"]) for l in self.stack_config])
        self.downsamples = nn.ModuleList([Downsample2() for _ in range(len(stack_config) - 1)])
        self.upsamples = nn.ModuleList([Upsample2() for _ in range(len(stack_config) - 1)])
        self.projections = nn.ModuleList([
                nn.Linear(stack_config[i+1]['embed_dim'], stack_config[i]['embed_dim'])
                for i in range(len(stack_config) - 1)])
        self.downward_projections = nn.ModuleList([
                nn.Linear(stack_config[i]['embed_dim'], stack_config[i+1]['embed_dim'])
                for i in range(len(stack_config) - 1)])

    def forward(self, x):
        skip_outputs = []
        for i in range(len(self.layers)):
            x = self.layers[i](x)
            if i < len(self.downsamples):
                skip_outputs.append(x)
                x = self.downsamples[i](x)
                x = self.downward_projections[i](x)
        
        for i in range(len(self.upsamples) - 1, -1, -1):
            x = self.upsamples[i](x)
            x = self.projections[i](x)
            x = x + skip_outputs[i]
        
        return x
````

````
import torch
import torch.nn as nn
import torch.nn.functional as F

# ScaledLinear
def ScaledLinear(
    in_features: int,
    out_features: int,
    bias: bool = True,
    initial_scale: float = 1.0,
) -> nn.Linear:
    linear = nn.Linear(in_features, out_features, bias)
    with torch.no_grad():
        linear.weight[:] *= initial_scale
        if linear.bias is not None:
            torch.nn.init.uniform_(linear.bias, -0.1 * initial_scale, 0.1 * initial_scale)
    return linear

# ScaledConv1d
def ScaledConv1d(*args, initial_scale: float = 1.0, **kwargs) -> nn.Conv1d:
    ans = nn.Conv1d(*args, **kwargs)
    with torch.no_grad():
        ans.weight[:] *= initial_scale
        if ans.bias is not None:
            torch.nn.init.uniform_(ans.bias, -0.1 * initial_scale, 0.1 * initial_scale)
    return ans

# DoubleSwish
class DoubleSwish(nn.Module):
    def forward(self, x):
        return x * torch.sigmoid(x - 1)
````

````
# BasicNorm
class BasicNorm(nn.Module):
    def __init__(self, num_channels: int, channel_dim: int = -1, eps: float = 0.25):
        super().__init__()
        self.num_channels = num_channels
        self.channel_dim = channel_dim
        self.eps = eps
        self.scale = nn.Parameter(torch.ones(num_channels))
        self.bias = nn.Parameter(torch.zeros(num_channels))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        rms = torch.sqrt(torch.mean(x ** 2, dim=self.channel_dim, keepdim=True) + self.eps)
        x = x / rms
        if self.channel_dim == -1:
            return x * self.scale + self.bias
        else:
            shape = [1] * x.ndim
            shape[self.channel_dim] = self.num_channels
            scale = self.scale.view(shape)
            bias = self.bias.view(shape)
            return x * scale + bias
````

````
# BalancerFunction
class BalancerFunction(torch.autograd.Function):
    @staticmethod
    def forward(ctx, x, channel_dim, min_positive, max_positive, max_abs):
        dims = torch.arange(x.dim()).tolist()
        dims.pop(channel_dim)
        pos_proportion = (x > 0).float().mean(dim=dims)
        max_abs_val = x.abs().amax(dim=dims)
        ctx.save_for_backward(x, pos_proportion, max_abs_val)
        ctx.channel_dim = channel_dim
        ctx.min_positive = min_positive
        ctx.max_positive = max_positive
        ctx.max_abs = max_abs
        return x

    @staticmethod
    def _calc_penalty(proportion, threshold, x, channel_dim, comparison_type='bigger'):
        mask = proportion < threshold if comparison_type == 'lower' else proportion > threshold
        penalty = threshold - proportion
        penalty = penalty * mask
        
        # Add unsqueeze for each dimension except channel_dim
       # for i in range(x.ndim):
        #    if i != channel_dim:
         #       penalty = penalty.unsqueeze(i if i < channel_dim else i - 1)
        # Unsqueeze before channel_dim
        for _ in range(channel_dim if channel_dim >= 0 else x.ndim + channel_dim):
            penalty = penalty.unsqueeze(0)
        # Unsqueeze after channel_dim  
        for _ in range(x.ndim - (channel_dim if channel_dim >= 0 else x.ndim + channel_dim) - 1):
            penalty = penalty.unsqueeze(-1)
        
        return penalty

    @staticmethod
    def backward(ctx, grad_output):
        x, pos_proportion, max_abs_val = ctx.saved_tensors
        #print(f'shape of x in balancer function {x.shape}')
        channel_dim = ctx.channel_dim
        #print(f'channel dim{channel_dim}')
        min_positive = ctx.min_positive
        max_positive = ctx.max_positive
        max_abs = ctx.max_abs

        shape = [1] * x.ndim
        shape[channel_dim] = -1

        penalty_min = BalancerFunction._calc_penalty(pos_proportion, min_positive, x, channel_dim, comparison_type='lower')
        penalty_max = BalancerFunction._calc_penalty(pos_proportion, max_positive, x, channel_dim)
        penalty_abs = BalancerFunction._calc_penalty(max_abs_val, max_abs, x, channel_dim)
        #print(f'shape of penalty_abs in balancer function {penalty_abs.shape}')
        penalty_abs = penalty_abs * (-torch.sign(x))

        grad_input = grad_output + penalty_min - penalty_max + penalty_abs
        return grad_input, None, None, None, None

# ActivationBalancer
class ActivationBalancer(nn.Module):
    def __init__(
        self,
        num_channels: int,
        channel_dim: int,
        min_positive: float = 0.05,
        max_positive: float = 0.95,
        max_abs: float = 10.0,
        min_prob: float = 0.01,
    ):
        super().__init__()
        self.num_channels = num_channels
        self.channel_dim = channel_dim
        self.min_positive = min_positive
        self.max_positive = max_positive
        self.max_abs = max_abs
        self.min_prob = min_prob

    def forward(self, x):
        x = BalancerFunction.apply(x, self.channel_dim, self.min_positive, self.max_positive, self.max_abs)
        return x
````

````
# WhitenFunction
class WhitenFunction(torch.autograd.Function):
    @staticmethod
    def forward(ctx, x, whitening_limit, grad_scale, num_groups):
        batch, seq_len, num_channels = x.shape
        channels_per_group = num_channels // num_groups
        x_reshaped = x.reshape(-1, num_groups, channels_per_group)
        x_whitened_groups = []
        L_clamped_list = []
        V_list = []
        eps = 1e-5
        for g in range(num_groups):
            x_group = x_reshaped[:, g, :]
            cov_mat = torch.cov(x_group.T)
            L, V = torch.linalg.eigh(cov_mat)
            L_clamped = torch.clamp(L, min=eps)
            L_clamped_list.append(L_clamped)
            V_list.append(V)
            L_whitened = (1 - whitening_limit) * L_clamped + whitening_limit * 1.0
            x_whitened = x_group @ V @ torch.diag(1.0/torch.sqrt(L_whitened)) @ V.T
            x_whitened = x_whitened.reshape(batch*seq_len, channels_per_group)
            x_whitened_groups.append(x_whitened)
        x_whitened_groups = torch.cat(x_whitened_groups, dim=1).reshape(batch, seq_len, num_channels)
        L_clamped_all = torch.stack(L_clamped_list)
        V_all = torch.stack(V_list)
        ctx.save_for_backward(x, L_clamped_all, V_all)
        ctx.num_groups = num_groups
        ctx.whitening_limit = whitening_limit
        ctx.grad_scale = grad_scale
        return x_whitened_groups

    @staticmethod
    def backward(ctx, grad_output):
        x, L_clamped_all, V_all = ctx.saved_tensors
        num_groups = ctx.num_groups
        grad_scale = ctx.grad_scale
        batch, seq_len, num_channels = x.shape
        channels_per_group = num_channels // num_groups
        grad_reshaped = grad_output.reshape(-1, num_groups, channels_per_group)
        grad_inputs = []
        for g in range(num_groups):
            grad_reshaped_group = grad_reshaped[:, g, :]
            grad_reshaped_group = grad_reshaped_group * grad_scale
            L_clamped_group = L_clamped_all[g, :]
            sqrt_L = torch.sqrt(L_clamped_group)
            inv_sqrt_L = 1.0 / sqrt_L
            V_group = V_all[g, :]
            grad_V = grad_reshaped_group @ V_group @ torch.diag(inv_sqrt_L)
            grad_input = grad_V @ V_group.T
            grad_input = grad_input.reshape(batch*seq_len, channels_per_group)
            grad_inputs.append(grad_input)
        grad_inputs = torch.cat(grad_inputs, dim=1).reshape(batch, seq_len, num_channels)
        return grad_inputs, None, None, None

# Whiten
class Whiten(nn.Module):
    def __init__(
        self,
        num_groups: int,
        whitening_limit: float,
        prob: float,
        grad_scale: float,
    ):
        super().__init__()
        self.num_groups = num_groups
        self.whitening_limit = whitening_limit
        self.prob = prob
        self.grad_scale = grad_scale

    def forward(self, x):
        if self.prob < torch.rand(1).item():
            return x
        x_whitened = WhitenFunction.apply(x, self.whitening_limit, self.grad_scale, self.num_groups)
        return x_whitened
````

````
# BypassModule
class BypassModule(nn.Module):
    def __init__(
        self,
        embed_dim: int,
        skip_rate: float = 0.0,
        straight_through_rate: float = 0.0,
        scale_min: float = 0.9,
        scale_max: float = 1.1,
    ):
        super().__init__()
        self.embed_dim = embed_dim
        self.skip_rate = skip_rate
        self.straight_through_rate = straight_through_rate
        self.scale_min = scale_min
        self.scale_max = scale_max
        self.scale = nn.Parameter(torch.ones(1))

    def forward(self, x):
        if self.training and torch.rand(1).item() < self.straight_through_rate:
            return x
        if self.training and torch.rand(1).item() < self.skip_rate:
            return torch.zeros_like(x)
        scale_clamped = torch.clamp(self.scale, self.scale_min, self.scale_max)
        return x * scale_clamped

# PoolingModule
class PoolingModule(nn.Module):
    def __init__(self, d_model, pooling_size):
        super().__init__()
        self.d_model = d_model
        self.pooling_size = pooling_size
        self.linear1 = ScaledLinear(d_model, d_model)
        self.linear2 = ScaledLinear(d_model, d_model)
        self.activation = DoubleSwish()

    def forward(self, x):
        x = self.linear1(x)
        x = self.activation(x)
        x = x.transpose(2, 1)
        x = F.avg_pool1d(x, self.pooling_size)
        x = x.transpose(2, 1)
        x = self.linear2(x)
        return x
````

````
# BiasNorm (from your original code)
class BiasNormFunction(torch.autograd.Function):
    @staticmethod
    def forward(ctx, x, bias, log_scale):
        x_debiased = x - bias
        var = torch.mean(x_debiased**2, dim=-1, keepdim=True)
        scale = var**(-0.5) * torch.exp(log_scale)
        output = x * scale
        ctx.save_for_backward(output, bias, scale, log_scale)
        return output

    @staticmethod
    def backward(ctx, grad_output):
        output, bias, scale, log_scale = ctx.saved_tensors
        x = output / scale
        x_debiased = x - bias
        var = torch.mean(x_debiased**2, dim=-1, keepdim=True)
        grad_log_scale = torch.sum(grad_output * output).view(1)
        N = x_debiased.shape[-1]
        common_term = torch.sum(grad_output * x, dim=-1, keepdim=True)
        grad_bias = torch.sum(common_term * scale * var**(-1) * (1/N) * x_debiased, dim=(0,1))
        grad_x = grad_output * scale - scale * var**(-1) * (1/N) * x_debiased * torch.sum(grad_output * x, dim=-1, keepdim=True)
        return grad_x, grad_bias, grad_log_scale

class BiasNorm(nn.Module):
    def __init__(self, num_channels):
        super().__init__()
        self.num_channels = num_channels
        self.bias = nn.Parameter(torch.empty(self.num_channels).uniform_(-0.01, 0.01))
        self.log_scale = nn.Parameter(torch.tensor(0.0))

    def forward(self, x):
        return BiasNormFunction.apply(x, self.bias, self.log_scale)

# RelPositionalEncoding
class RelPositionalEncoding(nn.Module):
    def __init__(self, embedding_dim, max_pos):
        super().__init__()
        self.embedding_dim = embedding_dim
        self.max_pos = max_pos
        self.emb = nn.Embedding(2*self.max_pos+1, embedding_dim)
    
    def forward(self, x):
        batch_n, seq_len, channels = x.shape
        t = torch.arange(seq_len)
        dist_mat = (t - t.unsqueeze(1)).clip(-self.max_pos, self.max_pos) + self.max_pos
        return self.emb(dist_mat)
````

````
# RelPositionMultiheadAttentionWeights
class RelPositionMultiheadAttentionWeights(nn.Module):
    def __init__(self, embed_dim, num_heads, query_head_dim, pos_dim, pos_head_dim, dropout=0.0):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.pos_dim = pos_dim
        self.pos_head_dim = pos_head_dim
        self.query_head_dim = query_head_dim
        self.dropout = nn.Dropout(dropout)
        self.in_proj = ScaledLinear(self.embed_dim, (2*self.num_heads*self.query_head_dim) + (self.num_heads*self.pos_dim))

    def forward(self, x, pos_emb):
        batch_n, seq_len, dim = x.shape
        x = self.in_proj(x)
        q = x[..., :self.num_heads*self.query_head_dim]
        k = x[..., self.num_heads*self.query_head_dim:2*self.num_heads*self.query_head_dim]
        p = x[..., 2*self.num_heads*self.query_head_dim:]
        
        q = q.reshape(batch_n, seq_len, self.num_heads, self.query_head_dim).transpose(1, 2)
        k = k.reshape(batch_n, seq_len, self.num_heads, self.query_head_dim).transpose(1, 2)
        p = p.reshape(batch_n, seq_len, self.num_heads, self.pos_dim).transpose(1, 2)
        
        pos_scores = torch.einsum('bhid,ijd->bhij', p, pos_emb)
        attn_scores = q @ k.transpose(-2, -1)
        attn_scores = attn_scores/(self.query_head_dim ** 0.5) + pos_scores
        attn_weights = torch.softmax(attn_scores, dim=-1)
        attn_weights = self.dropout(attn_weights)
        return attn_weights

# SelfAttention
class SelfAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, value_head_dim):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.value_head_dim = value_head_dim
        self.proj1 = ScaledLinear(self.embed_dim, self.num_heads * self.value_head_dim)
        self.proj2 = ScaledLinear(self.num_heads * self.value_head_dim, self.embed_dim)

    def forward(self, x, attn_weights):
        x = self.proj1(x)
        batch_n, seq_len, _ = x.shape
        x = x.reshape((batch_n, seq_len, self.num_heads, self.value_head_dim)).transpose(1, 2)
        x = attn_weights @ x
        x = x.transpose(1, 2).reshape((batch_n, seq_len, -1))
        x = self.proj2(x)
        return x

# NonLinearAttention
class NonLinearAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, head_dim):
        super().__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = head_dim
        self.proj_value1 = ScaledLinear(self.embed_dim, self.num_heads * self.head_dim)
        self.proj_value2 = ScaledLinear(self.num_heads * self.head_dim, self.embed_dim)
        self.proj_query1 = ScaledLinear(self.embed_dim, self.num_heads * self.head_dim)
        self.activation_fun = lambda x: torch.log(1+torch.exp(x-1))-0.08*x-0.313261687

    def forward(self, x, attn_weights):
        value = self.proj_value1(x)
        query = self.proj_query1(x)
        batch_n, seq_len, _ = value.shape
        value = value.reshape((batch_n, seq_len, self.num_heads, self.head_dim)).transpose(1, 2)
        query = query.reshape((batch_n, seq_len, self.num_heads, self.head_dim)).transpose(1, 2)
        output = self.activation_fun(query) * (attn_weights @ value)
        output = output.transpose(1, 2).reshape((batch_n, seq_len, -1))
        output = self.proj_value2(output)
        return output
````

````
# FeedforwardModule
class FeedforwardModule(nn.Module):
    def __init__(self, embed_dim, feedforward_dim, dropout):
        super().__init__()
        self.embed_dim = embed_dim
        self.feedforward_dim = feedforward_dim
        self.dropout = dropout
        self.linear1 = nn.Linear(self.embed_dim, self.feedforward_dim)
        self.activation_fun = lambda x: torch.log(1+torch.exp(x-4))-0.08*x-0.035
        self.drop = nn.Dropout(self.dropout)
        self.linear2 = nn.Linear(self.feedforward_dim, self.embed_dim)

    def forward(self, x):
        x = self.linear1(x)
        x = self.activation_fun(x)
        x = self.drop(x)
        x = self.linear2(x)
        return x

# ConvNeXt
class ConvNeXt(nn.Module):
    def __init__(self, embed_dim, kernel_size, expansion_factor, dropout):
        super().__init__()
        self.embed_dim = embed_dim
        self.kernel_size = kernel_size
        self.expansion_factor = expansion_factor
        self.dropout_n = dropout
        
        self.depth_conv = nn.Conv1d(self.embed_dim, self.embed_dim, self.kernel_size, 
                                     padding=self.kernel_size // 2, groups=self.embed_dim)
        self.point_expansion = nn.Linear(self.embed_dim, self.embed_dim*self.expansion_factor)
        self.activation_fun = lambda x: torch.log(1+torch.exp(x-4))-0.08*x-0.035
        self.dropout = nn.Dropout(self.dropout_n)
        self.point_compression = nn.Linear(self.embed_dim*self.expansion_factor, self.embed_dim)

    def forward(self, x):
        x = torch.transpose(x, 2, 1)
        x = self.depth_conv(x)
        x = torch.transpose(x, 2, 1)
        x = self.point_expansion(x)
        x = self.activation_fun(x)
        x = self.dropout(x)
        x = self.point_compression(x)
        return x
````

````
# ZipformerBlock
class ZipformerBlock(nn.Module):
    def __init__(self, embed_dim, num_heads, query_head_dim, value_head_dim, 
                 pos_dim, pos_head_dim, feedforward_dim, kernel_size, expansion_factor, 
                 dropout=0.1, pooling_size=2):
        super().__init__()
        self.embed_dim = embed_dim
        
        # Three feedforward modules (macaron style)
        self.ffn1 = FeedforwardModule(embed_dim, feedforward_dim, dropout)
        self.ffn2 = FeedforwardModule(embed_dim, feedforward_dim, dropout)
        self.ffn3 = FeedforwardModule(embed_dim, feedforward_dim, dropout)
        
        # Pooling module
        self.pooling = PoolingModule(embed_dim, pooling_size)
        
        # Attention
        self.attn_weights = RelPositionMultiheadAttentionWeights(
            embed_dim, num_heads, query_head_dim, pos_dim, pos_head_dim, dropout
        )
        self.sa = SelfAttention(embed_dim, num_heads, value_head_dim)
        self.nla = NonLinearAttention(embed_dim, num_heads, value_head_dim)
        
        # Two conv modules
        self.conv1 = ConvNeXt(embed_dim, kernel_size, expansion_factor, dropout)
        self.conv2 = ConvNeXt(embed_dim, kernel_size, expansion_factor, dropout)
        
        # Normalization
        self.norm1 = BiasNorm(embed_dim)
        self.norm2 = BiasNorm(embed_dim)
        self.norm3 = BiasNorm(embed_dim)
        self.norm_final = BasicNorm(embed_dim)
        
        # Balancer and Whiten
        self.balancer = ActivationBalancer(embed_dim, channel_dim=-1)
        self.whiten = Whiten(num_groups=1, whitening_limit=0.5, prob=0.5, grad_scale=0.1)
        
        # Bypass modules for skip connections
        self.bypass1 = BypassModule(embed_dim)
        self.bypass2 = BypassModule(embed_dim)

    def forward(self, x, pos_emb):
        x = x + self.bypass1(self.ffn1(self.norm1(x)))
        attn_weights = self.attn_weights(x, pos_emb)
        x = x + self.sa(self.norm2(x), attn_weights)
        x = x + self.conv1(x)
        x = x + self.ffn2(x)
        x = x + self.nla(self.norm3(x), attn_weights)
        x = x + self.conv2(x)
        x = x + self.ffn3(x)
        x = self.norm_final(x)
        x = self.balancer(x)
        x = self.whiten(x)
        return x

# ZipformerEncoder
class ZipformerEncoder(nn.Module):
    def __init__(self, num_layers, embed_dim=64, num_heads=4, query_head_dim=16, value_head_dim=16,
                 pos_dim=32, pos_head_dim=8, feedforward_dim=256, kernel_size=7,
                 expansion_factor=2, dropout=0.1, embedding_dim=8, max_pos=32, pooling_size=2):
        super().__init__()
        self.num_layers = num_layers
        self.embed_dim = embed_dim
        self.pos_encoder = RelPositionalEncoding(embedding_dim=embedding_dim, max_pos=max_pos)
        self.layers = nn.ModuleList([ZipformerBlock(
                            embed_dim=embed_dim, num_heads=num_heads, query_head_dim=query_head_dim, 
                            value_head_dim=value_head_dim, pos_dim=pos_dim, pos_head_dim=pos_head_dim, 
                            feedforward_dim=feedforward_dim, kernel_size=kernel_size,
                            expansion_factor=expansion_factor, dropout=dropout,
                            pooling_size=pooling_size
                            ) for i in range(self.num_layers)])

    def forward(self, x):
        pos_emb = self.pos_encoder(x)
        for layer in self.layers:
            x = layer(x, pos_emb)
        return x
````

````
# DownsampledZipformerEncoder
class DownsampledZipformerEncoder(nn.Module):
    def __init__(self, encoder, input_dim, output_dim, downsample):
        super().__init__()
        self.downsample = downsample
        self.encoder = encoder
        self.input_dim = input_dim
        self.output_dim = output_dim
        self.input_proj = ScaledLinear(input_dim, encoder.embed_dim)
        self.output_proj = ScaledLinear(encoder.embed_dim, output_dim)

    def forward(self, x):
        x = x.transpose(2, 1)
        x = F.avg_pool1d(x, kernel_size=self.downsample)
        x = x.transpose(2, 1)
        x = self.input_proj(x)
        x = self.encoder(x)
        x = self.output_proj(x)
        return x

# SimpleCombiner
class SimpleCombiner(nn.Module):
    def __init__(self, num_inputs, embed_dim):
        super().__init__()
        self.num_inputs = num_inputs
        self.embed_dim = embed_dim
        self.weights = nn.Parameter(torch.ones(num_inputs))

    def forward(self, x):
        normalized_weights = F.softmax(self.weights, dim=0).unsqueeze(1).unsqueeze(2).unsqueeze(3)
        # Find minimum size
        min_len = min(t.shape[1] for t in x)
        # Trim all to min_len
        x = [t[:, :min_len, :] for t in x]
        x = torch.stack(x)
        output = torch.sum(x * normalized_weights, dim=0)
        return output
````

````
class ConvEmbed(nn.Module):
    def __init__(self, input_channels, output_dim):
        super().__init__()
        self.conv1 = nn.Conv2d(input_channels, 8, kernel_size=3, stride=(1,2), padding=1)
        self.act1 = DoubleSwish()
        self.balancer1 = ActivationBalancer(8, channel_dim=1)
        
        self.conv2 = nn.Conv2d(8, 32, kernel_size=3, stride=(2,2), padding=1)
        self.act2 = DoubleSwish()
        self.balancer2 = ActivationBalancer(32, channel_dim=1)
        
        self.conv3 = nn.Conv2d(32, 128, kernel_size=3, stride=(1,2), padding=1)
        self.act3 = DoubleSwish()
        self.balancer3 = ActivationBalancer(128, channel_dim=1)
        
        self.convnext = ConvNeXt(1280, kernel_size=7, expansion_factor=3, dropout=0.0)
        self.linear = ScaledLinear(1280, output_dim)
        self.norm = BiasNorm(output_dim)
    
    def forward(self, x):
        x = x.unsqueeze(1)
        x = self.balancer1(self.act1(self.conv1(x)))
        x = self.balancer2(self.act2(self.conv2(x)))
        x = self.balancer3(self.act3(self.conv3(x)))
        batch, _, time, _ = x.shape
        x = x.permute(0, 2, 1, 3).reshape(batch, time, -1)
        x = self.convnext(x)
        x = self.linear(x)
        x = self.norm(x)
        return x
````

````
class MultiScaleZipformerEncoder(nn.Module):
    def __init__(self, input_dim=80):
        super().__init__()
        
        # Input projection
        #self.input_proj = ScaledLinear(input_dim, 128)
        self.conv_embed = ConvEmbed(input_channels=1, output_dim=128)
        
        # Stack 0: 128 dim, full resolution
        self.encoder0 = ZipformerEncoder(
            num_layers=2, embed_dim=128, num_heads=4,
            query_head_dim=32, value_head_dim=32,
            pos_dim=32, pos_head_dim=8,
            feedforward_dim=512, kernel_size=7,
            expansion_factor=2, dropout=0.1,
            embedding_dim=32, max_pos=200
        )
        
        # Stack 1: 256 dim, downsample by 2
        encoder1 = ZipformerEncoder(
            num_layers=2, embed_dim=256, num_heads=4,
            query_head_dim=64, value_head_dim=64,
            pos_dim=32, pos_head_dim=8,
            feedforward_dim=1024, kernel_size=7,
            expansion_factor=2, dropout=0.1,
            embedding_dim=32, max_pos=100
        )
        self.encoder1 = DownsampledZipformerEncoder(encoder1, input_dim=128, output_dim=256, downsample=2)
        
        # Stack 2: 512 dim, downsample by 2
        encoder2 = ZipformerEncoder(
            num_layers=2, embed_dim=512, num_heads=8,
            query_head_dim=64, value_head_dim=64,
            pos_dim=32, pos_head_dim=8,
            feedforward_dim=2048, kernel_size=7,
            expansion_factor=2, dropout=0.1,
            embedding_dim=32, max_pos=50
        )
        self.encoder2 = DownsampledZipformerEncoder(encoder2, input_dim=256, output_dim=512, downsample=2)
        
        # Upsampling projections
        self.upsample1_proj = ScaledLinear(512, 256)
        self.upsample0_proj = ScaledLinear(256, 128)
        
        # Combiners for skip connections
        self.combiner1 = SimpleCombiner(num_inputs=2, embed_dim=256)
        self.combiner0 = SimpleCombiner(num_inputs=2, embed_dim=128)
        
    def forward(self, x):
        # Input projection
        x = self.conv_embed(x)
        
        # Downward path with skip connections
        x0 = self.encoder0(x)
        x1 = self.encoder1(x0)
        x2 = self.encoder2(x1)
        
        # Upward path with skip connections
        # Upsample from 512 to 256
        x_up1 = F.interpolate(x2.transpose(1, 2), scale_factor=2, mode='linear').transpose(1, 2)
        x_up1 = self.upsample1_proj(x_up1)
        x1_combined = self.combiner1([x1, x_up1])
        
        # Upsample from 256 to 128
        x_up0 = F.interpolate(x1_combined.transpose(1, 2), scale_factor=2, mode='linear').transpose(1, 2)
        x_up0 = self.upsample0_proj(x_up0)
        x0_combined = self.combiner0([x0, x_up0])
        
        return x0_combined
````

````
import torchaudio
from torch.utils.data import Dataset, DataLoader

# Feature extraction
mel_spectrogram = torchaudio.transforms.MelSpectrogram(
    sample_rate=16000, n_fft=400, hop_length=160, n_mels=80
)

# ASR Model with CTC head
class ZipformerASR(nn.Module):
    def __init__(self, encoder, vocab_size, output_dim):
        super().__init__()
        self.encoder = encoder
        self.ctc_head = nn.Linear(output_dim, vocab_size)
    
    def forward(self, x):
        x = x.transpose(1, 2)
        x = self.encoder(x)
        logits = self.ctc_head(x)
        return logits

# Vocabulary
vocab = list("abcdefghijklmnopqrstuvwxyz .,")
vocab_size = len(vocab)
char_to_idx = {c: i for i, c in enumerate(vocab)}

# Dataset
class ASRDataset(Dataset):
    def __init__(self, audio_files, transcriptions):
        self.audio_files = audio_files
        self.transcriptions = transcriptions
        
    def __len__(self):
        return len(self.audio_files)
    
    def __getitem__(self, idx):
        waveform, sr = torchaudio.load(self.audio_files[idx])
        if waveform.shape[0] > 1:
            waveform = torch.mean(waveform, dim=0, keepdim=True)
        
        mel_features = mel_spectrogram(waveform)
        text = self.transcriptions[idx]
        encoded = torch.tensor([char_to_idx[c] for c in text if c in char_to_idx])
        
        return mel_features.squeeze(0), encoded

def asr_collate_fn(batch):
    mel_features, transcriptions = zip(*batch)
    
    mel_lengths = [m.shape[1] for m in mel_features]
    max_mel_len = max(mel_lengths)
    padded_mels = torch.stack([F.pad(m, (0, max_mel_len - m.shape[1])) for m in mel_features])
    
    text_lengths = [t.shape[0] for t in transcriptions]
    max_text_len = max(text_lengths)
    padded_texts = torch.stack([F.pad(t, (0, max_text_len - t.shape[0])) for t in transcriptions])
    
    return padded_mels, torch.tensor(mel_lengths), padded_texts, torch.tensor(text_lengths)

# Training function
def train_asr(model, dataloader, num_epochs=5, lr=1e-4):
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    ctc_loss = nn.CTCLoss(blank=0, zero_infinity=True)
    model.train()
    
    for epoch in range(num_epochs):
        total_loss = 0
        for batch_idx, (mel_features, mel_lens, targets, target_lens) in enumerate(dataloader):
            optimizer.zero_grad()
            
            logits = model(mel_features)
            log_probs = F.log_softmax(logits, dim=-1)
            log_probs = log_probs.transpose(0, 1)
            
            input_lengths = mel_lens // 2
            #print(f"log_probs shape: {log_probs.shape}")
            #print(f"input_lengths: {input_lengths}")
            #print(f"target_lens: {target_lens}")
            #print(f"Has NaN in log_probs: {torch.isnan(log_probs).any()}")
            loss = ctc_loss(log_probs, targets, input_lengths, target_lens)
            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            print(f"Epoch {epoch}, Batch {batch_idx}, Loss: {loss.item():.4f}")
        
        avg_loss = total_loss / len(dataloader)
        print(f"Epoch {epoch} complete. Average Loss: {avg_loss:.4f}")
    
    return model

# Inference function
def transcribe(model, audio_file):
    model.eval()
    with torch.no_grad():
        waveform, sr = torchaudio.load(audio_file)
        if waveform.shape[0] > 1:
            waveform = torch.mean(waveform, dim=0, keepdim=True)
        
        mel_features = mel_spectrogram(waveform)
        mel_features = mel_features.squeeze(0).unsqueeze(0)
        
        logits = model(mel_features)
        predictions = torch.argmax(logits, dim=-1).squeeze(0)
        
        decoded = []
        prev_token = None
        for token in predictions:
            token = token.item()
            if token != 0 and token != prev_token:
                decoded.append(vocab[token])
            prev_token = token
        
        return ''.join(decoded)
````

````
import random
import torchaudio
from torch.utils.data import Dataset, DataLoader

# Feature extraction
mel_spectrogram = torchaudio.transforms.MelSpectrogram(
    sample_rate=16000, n_fft=400, hop_length=160, n_mels=80
)

def asr_collate_fn(batch):
    mel_features, transcriptions = zip(*batch)
    
    mel_lengths = [m.shape[1] for m in mel_features]
    max_mel_len = max(mel_lengths)
    padded_mels = torch.stack([F.pad(m, (0, max_mel_len - m.shape[1])) for m in mel_features])
    
    text_lengths = [t.shape[0] for t in transcriptions]
    max_text_len = max(text_lengths)
    padded_texts = torch.stack([F.pad(t, (0, max_text_len - t.shape[0])) for t in transcriptions])
    
    return padded_mels, torch.tensor(mel_lengths), padded_texts, torch.tensor(text_lengths)

# Generate synthetic dataset
def create_synthetic_dataset(num_samples=100):
    audio_data = []
    labels = []
    for i in range(num_samples):
        mel_features = torch.randn(80, 200)
        text = ''.join(random.choices('abcdefghijklmnopqrstuvwxyz ', k=random.randint(5, 15)))
        audio_data.append(mel_features)
        labels.append(text)
    return audio_data, labels

# Synthetic Dataset class
class SyntheticASRDataset(Dataset):
    def __init__(self, audio_data, labels):
        self.audio_data = audio_data
        self.labels = labels
        
    def __len__(self):
        return len(self.audio_data)
    
    def __getitem__(self, idx):
        mel_features = self.audio_data[idx]
        text = self.labels[idx]
        encoded = torch.tensor([char_to_idx[c] for c in text if c in char_to_idx])
        return mel_features, encoded

class ZipformerASR(nn.Module):
    def __init__(self, encoder, vocab_size, output_dim):
        super().__init__()
        self.encoder = encoder
        self.ctc_head = nn.Linear(output_dim, vocab_size)
    
    def forward(self, x):
        x = x.transpose(1, 2)
        x = self.encoder(x)
        logits = self.ctc_head(x)
        return logits

def train_asr(model, dataloader, num_epochs=5, lr=1e-4):
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr)
    ctc_loss = nn.CTCLoss(blank=0, zero_infinity=True)
    model.train()
    
    for epoch in range(num_epochs):
        total_loss = 0
        for batch_idx, (mel_features, mel_lens, targets, target_lens) in enumerate(dataloader):
            optimizer.zero_grad()
            
            logits = model(mel_features)
            log_probs = F.log_softmax(logits, dim=-1)
            log_probs = log_probs.transpose(0, 1)
            
            input_lengths = mel_lens // 2
            #print(f"log_probs shape: {log_probs.shape}")
            #print(f"input_lengths: {input_lengths}")
            #print(f"target_lens: {target_lens}")
            #print(f"Has NaN in log_probs: {torch.isnan(log_probs).any()}")
            loss = ctc_loss(log_probs, targets, input_lengths, target_lens)
            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            print(f"Epoch {epoch}, Batch {batch_idx}, Loss: {loss.item():.4f}")
        
        avg_loss = total_loss / len(dataloader)
        print(f"Epoch {epoch} complete. Average Loss: {avg_loss:.4f}")
    
    return model

# Create dataset
audio_data, labels = create_synthetic_dataset(num_samples=100)
synthetic_dataset = SyntheticASRDataset(audio_data, labels)
synthetic_dataloader = DataLoader(synthetic_dataset, batch_size=4, collate_fn=asr_collate_fn)

# Recreate encoder with Whiten disabled
#encoder = ZipformerEncoder(
 #   num_layers=2, embed_dim=80, num_heads=4,
 #   query_head_dim=20, value_head_dim=20,
 #   pos_dim=32, pos_head_dim=8,
 #   feedforward_dim=256, kernel_size=7,
 #   expansion_factor=2, dropout=0.1,
 #   embedding_dim=32, max_pos=200
#)
# Vocabulary
vocab = list("abcdefghijklmnopqrstuvwxyz .,")
vocab_size = len(vocab)
char_to_idx = {c: i for i, c in enumerate(vocab)}

encoder=MultiScaleZipformerEncoder(input_dim=80)
# Update ZipformerBlock to disable Whiten first!
asr_model = ZipformerASR(encoder, vocab_size, output_dim=128)
````

````
def transcribe(model, audio_file):
    model.eval()
    with torch.no_grad():
        waveform, sr = torchaudio.load(audio_file)
        if waveform.shape[0] > 1:
            waveform = torch.mean(waveform, dim=0, keepdim=True)
        
        mel_features = mel_spectrogram(waveform)
        mel_features = mel_features.squeeze(0).unsqueeze(0)
        
        logits = model(mel_features)
        predictions = torch.argmax(logits, dim=-1).squeeze(0)
        
        decoded = []
        prev_token = None
        for token in predictions:
            token = token.item()
            if token != 0 and token != prev_token:
                decoded.append(vocab[token])
            prev_token = token
        
        return ''.join(decoded)
````

**The `exp(x - 1)` part:**
- Shifts the exponential to the right by 1
- Controls where the "knee" of the curve happens

**The `+ 1` part:**
- Ensures the value inside the log is always > 1
- Prevents `log(0)` which would be undefined
- Gives a minimum output value

Together, `log(1 + exp(x - 1))` is the **softplus function shifted right by 1**.

Does this help clarify the formula structure?

-4 shifts exponential by 4?

##### 🤖Reply🤖<!-- SOLVEIT_SEPARATOR_7f3a9b2c -->

Yes, exactly! `exp(x - 4)` shifts the exponential curve **4 units to the right**.

This means:
- The "knee" (where it starts to curve up) happens at x = 4 instead of x = 0
- For x < 4, the exponential is very small
- For x > 4, it grows rapidly

That's why Swooshl (with x-4) stays flatter for longer in the negative region compared to Swooshr (with x-1).
